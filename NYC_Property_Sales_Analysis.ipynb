```python
import pandas as pd
import numpy as np
#import matplotlib.pyplot as plt
#import seaborn as sns
```

```python
url = "https://data.cityofnewyork.us/api/v3/views/w2pb-icbu/query.csv"

df = pd.read_csv(url)

print(df.shape)
print(df.columns)
df.head()
```

```python
df.info()
```

```python
missing = df.isnull().sum()
missing[missing > 0].sort_values(ascending=False)
```

```python
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
missing_pct[missing_pct > 0].sort_values(ascending=False)
```

```python
df['sale_price'].describe()
```

```python
df['sale_price'].value_counts().head(20)
```

```python
(df['sale_price'] <= 0).sum()
```

```python
zero_sales = df[df['sale_price'] == 0]
zero_sales['building_class_category'].value_counts().head(20)
```

```python
ten_sales = df[df['sale_price'] == 10]
ten_sales['building_class_category'].value_counts().head(20)
```

```python
print(zero_sales['sale_date'].min(), zero_sales['sale_date'].max())
print(ten_sales['sale_date'].min(), ten_sales['sale_date'].max())
```

```python
clean_df = df[df['sale_price'] > 10].copy()
print(clean_df.shape)
print(clean_df['sale_price'].describe())
print((clean_df['sale_price'] <= 10).sum())
```

```python
q1 = np.percentile(clean_df['sale_price'], 25)
q3 = np.percentile(clean_df['sale_price'], 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)

upper_limit = q3 + 1.5 * iqr

print("Upper limit:", upper_limit)
```

```python
outliers = clean_df[clean_df['sale_price'] > upper_limit]
print(outliers.shape)
```

```python
outliers['borough'].value_counts()
```

```python
df.groupby('borough')['neighborhood'].unique()
```

```python
borough_map = {
    1: 'Manhattan',
    2: 'Bronx',
    3: 'Brooklyn',
    4: 'Queens',
    5: 'Staten Island'
}

clean_df['borough_name'] = clean_df['borough'].map(borough_map)
```

```python
clean_df[['borough', 'borough_name']].drop_duplicates().sort_values('borough')
```

```python
outliers['building_class_category'].value_counts().head(15)
```

```python
clean_df['building_class_category'].str.strip().value_counts().head(15)
```

```python
clean_df['building_class_category'].str.strip().str.replace(r'\s+', ' ', regex=True).value_counts()
```

```python
clean_df['building_class_category'] = (
    clean_df['building_class_category']
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
)
clean_df['building_class_category'].value_counts().head(15)
```

```python
outliers = clean_df[clean_df['sale_price'] > upper_limit].copy()
outliers.shape
```

```python
outliers.sort_values(
    'sale_price',
    ascending=False
)[[
    'sale_price',
    'borough_name',
    'neighborhood',
    'building_class_category',
    'address'
]].head(20)
```

```python
clean_df['sale_date'] = pd.to_datetime(clean_df['sale_date'])
clean_df['sale_date'].dtype
```

```python
clean_df['gross_square_feet'] = pd.to_numeric(
    clean_df['gross_square_feet'].str.replace(',', ''),
    errors='coerce'
)
```

```python
print("Land square feet = 0:", (clean_df['land_square_feet'] == 0).sum())
print("Land square feet missing:", clean_df['land_square_feet'].isna().sum())

print("Gross square feet = 0:", (clean_df['gross_square_feet'] == 0).sum())
print("Gross square feet missing:", clean_df['gross_square_feet'].isna().sum())
```

```python
sql_df = clean_df[[
    'borough',
    'borough_name',
    'neighborhood',
    'building_class_category',
    'residential_units',
    'commercial_units',
    'total_units',
    'land_square_feet',
    'gross_square_feet',
    'year_built',
    'sale_price',
    'sale_date',
    'zip_code',
    'latitude',
    'longitude',
    'nta'
]].copy()

sql_df.shape
```

```python
mysql_df = sql_df.copy()

numeric_cols = [
    'residential_units',
    'commercial_units',
    'total_units',
    'land_square_feet',
    'gross_square_feet',
    'year_built',
    'zip_code',
    'latitude',
    'longitude'
]

print(mysql_df.shape)
mysql_df.dtypes
```

```python
(mysql_df[numeric_cols] == -1).sum()
```

```python
mysql_df[numeric_cols] = mysql_df[numeric_cols].fillna(-1)
mysql_df.to_csv(
    'nyc_property_sales_mysql.csv',
    index=False,
)
```
